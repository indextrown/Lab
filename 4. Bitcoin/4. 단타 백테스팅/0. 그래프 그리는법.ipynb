### range vs nparange

### range
- 실제 숫자를 저장하지 않고 필요시 값을 계산해 제공하는 지연(lazy) 객체
- 정수만 생성 가능
- 메모리 거의 사용하지 않지만 Python 레벨 반복문 기반이라 연산속도 느림

### np.arange
- np.arange는 메모리 연속 영역에 동일 타입으로 ndarray 생성
- 모든 값이 실제 메모리에 존재
- 정수/실수 모두 생성 가능
- 내부가 C기반 벡터/행렬 연산
- 배열 크기만큼 메모리를 사용하지만 연산 속도 매우 빠름

```python
import numpy as np
import time

# range 순회
start = time.time()
for x in range(10 ** 5): x * 2
print(time.time() - start)

# np.arrange 순회
start = time.time()
for x in np.arange(10 ** 5): x * 2
print(time.time() - start)

# np.arange 벡터 연산
start = time.time()
np.arange(10 ** 5) * 2
print(time.time() - start)
```

### 기본 그래프

```python
from matplotlib import pyplot as plt
import numpy as np

x = np.arange(1, 10) 
y = x*5

plt.plot(x, y)
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("graph sample")
plt.show()
```

### 레전드(그래프를 여러 개)

```python
from matplotlib import pyplot as plt
import numpy as np

x = np.arange(1, 10, 0.1) 
y = x*0.2
y2 = np.sin(x)

plt.plot(x, y, 'b', label='first')
plt.plot(x, y2, 'r', label='second')
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.title("graph sample")
plt.legend(loc='upper right')
plt.show()
```

### 그래프 2개

```python
from matplotlib import pyplot as plt
import numpy as np

x = np.arange(1, 10) 
y = x*5

# 1행 2열(세로로배치할거면 2행 1열로) / fig는 객체를 의미 ax는 개별 그래프 배열을 의미 / figsize는 가로, 세로 인치 지정
fig, ax = plt.subplots(1, 2, figsize=(10, 4)) 

# 첫 번째 그래프
ax[0].plot(x, y, marker='o', color='b', linestyle="None")
ax[0].set_title("Graph 1")

# 두 번째 그래프
ax[1].plot(x, y, marker='o', color='r', linestyle="None")
ax[1].set_title("Graph 2")
```
